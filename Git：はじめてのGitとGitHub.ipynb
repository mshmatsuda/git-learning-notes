## 📂 セクション1〜2：Gitの基本概念と仕組み

💡 学んだ重要なポイント（メモ欄）

* Gitとは： ファイルの「変更履歴」を記録・管理するツール。
* コミット（Commit）： セーブポイントを作ること。いつでもこの状態に戻せる。
* リポジトリ（Repository）： 履歴を保存する「貯蔵庫（フォルダ）」。
* ローカルとリモート： 自分の作業環境（ローカル）と、ネット上の保管庫（リモート）。

## ✅ 実践チェックリスト

* 「ファイルのバージョンを増やす（ファイル名_ver2.0.xlsxなど）」という従来のやり方の何が問題なのか、人に説明できる。
* 「コミット」と「ただの上書き保存」の違いがイメージできる。
* コマンドのチャプターは、操作を完璧に真似しようとせず、動画を止めずにストーリーとして最後まで見られた。

## ☁️ セクション3〜4：GitHubの登録とリモートリポジトリ

💡 学んだ重要なポイント（メモ欄）

* GitHubとは： Gitで管理している履歴を、ネット上で共有・公開できるサービス。
* プッシュ（Push）： ローカルのセーブポイント（コミット）を、ネット上（GitHub）にアップロードすること。

## ✅ 実践チェックリスト

* GitHubのアカウントを無事に作成できた。
* 講座の通りに、GitHub上で最初の「リモートリポジトリ（空のフォルダ）」を作成できた。
* （★超重要・Colab連携）Google Colabを開き、メニューの「ファイル」＞「GitHubにコピーを保存」から、作成したリポジトリにノートブック（.ipynb）を保存できた。
* 自分のGitHubのページ（ブラウザ）を開き、Colabから送ったファイルが綺麗に表示されていることを確認した。

## 👥 セクション5〜：チーム開発の基礎（ブランチと競合）

💡 学んだ重要なポイント（メモ欄）

* ブランチ（Branch）： 履歴の流れを「枝分かれ」させる機能。本番環境を汚さずに実験できる。
* プルリクエスト（Pull Request）： 「変更したから確認して取り込んで！」と仲間に依頼すること。
* 競合（Conflict）： 同じ箇所のコードが同時に書き換えられて、Gitが混乱した状態。

## ✅ 実践チェックリスト

* なぜ「ブランチ」を作る必要があるのか（いきなり本番のコードを書き換えると何が危険か）が理解できた。
* クライアントから「プルリク（PR）を送ってください」と言われたときに、それが「コードのレビュー依頼のことだな」とピンとくる。
* コンフリクト（競合）が起きたとき、慌てずに「重なった部分をどちらかに統一して直せばいいんだな」と心構えができた。
